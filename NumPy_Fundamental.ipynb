# What is Numpy ?

### It is the fundamental package for scientific computing in Python. 
### At the core of the NumPy package , is the mdarray object.

## Numpy Arrays vs Python Sequences 
#### - Numpy arrays have a fixed size at created, unlile python list (which can grwo dynamically). Changing the size of an ndarray will create a new array and delete the original.

#### - The elements in a Numpy array are all required to be of the same data type, and thus will be the same size in memory.

#### - NumPy arrays faciliate advanced mathematical and other types of operations on large numbers of data. Typically, such operations are executed more eficiently and with less code than is possible using python's build in sequences.

# Array Attributes 


```python
a1=np.arange(10)
a2=np.arange(12,dtype=float).reshape(3,4)
a3=np.arange(8).reshape(2,2,2)
```

```python
# ndim : dimension of ndarray
print(a1.ndim)
print(a2.ndim)
print(a3.ndim)
```

```python
# shape : dimension item
print(a1.shape)
print(a2.shape)
print(a3.shape)
```

```python
# size : total element
print(a1.size)
print(a2.size)
print(a3.size)
```

```python
# itemsize : memory size occupied
# 64 bit is default
print(a1.itemsize)
print(a2.itemsize)
print(a3.itemsize)
```

```python
# dtype : data type of ndarray

print(a1.dtype)
print(a2.dtype)
print(a3.dtype)
```

# Changing Data type


```python
# astype : changing the data type 
print(a3.astype(np.int32))
```

# Array Functions 

```python
a1=np.random.random((3,3))
a1=np.round(a1*100)
print(a1)

```

```python
# max / min / sum / prod

print(np.max(a1))
print(np.min(a1))
print(np.sum(a1))
print(np.prod(a1))

# With axis
# 0 -> column and 1 -> row

print(np.max(a1,axis=0))
print(np.min(a1,axis=1))
print(np.sum(a1,axis=0))
print(np.prod(a1,axis=1))

```

```python
# mean/median/std/var
print(np.mean(a1))
print(np.median(a1))
print(np.std(a1))
print(np.var(a1))
```

```python
# dot product 
# np.dot()
a2=np.arange(12).reshape(3,4)
a3=np.arange(12,24).reshape(4,3)

np.dot(a2,a3)
```

```python
# log and exponents 
# np.log() and np.exp()
print(np.log(a1))
print(np.exp(a1))
```

```python
# round/floor/ceil

print(np.round(np.random.random((2,3))*100))
print(np.floor(np.random.random((2,3))*100))
print(np.ceil(np.random.random((2,3))*100))
```

# Indexing 



```python
a1=np.arange(10)
a2=np.arange(12).reshape(3,4)
a3=np.arange(27).reshape(3,3,3)
print(a1)
print(a2)
print(a3)
```

```python
# a1 : 4
print(a1[4]) # [index]
```

```python
# a2 : 11
print(a2[2,3]) # [row,column] : index starts with 0
```

```python
# a3 : 5
print(a3[1,0,1]) #[2D_index,row,colum]
```

# Silicing

# Stacking 


```python
# horizontal stacking 

# np.hstack(tuple)
a4=np.arange(12).reshape(3,4)
a5=np.arange(0,30).reshape(3,10)

print(a4)
print(a5)

np.hstack((a4,a5))
```

```python
# vertical stacking 

# np.vstack 

a4=np.arange(12).reshape(3,4)
a5=np.arange(0,4).reshape(1,4)

print(a4)
print(a5)

np.vstack((a4,a5))
```

# Splitting 
Splitting the array either in horizontal or vertical 

```python
# Horizontal Splitting 

# np.hsplit(array,no.of parts)
print(a4)

print(np.hsplit(a2,2))
```

```python
# Horizontal Splitting 

# np.vsplit(array,no.of parts)
print(a4)

print(np.vsplit(a2,3))
```
