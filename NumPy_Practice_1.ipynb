#### Create a NumPy array containing integers from 10 to 100 (inclusive), with a step of 10.

Then print:

#### NumPy Broadcasting Problem

```python
# Given 
A = np.array([[10, 20, 30],
              [40, 50, 60],
              [70, 80, 90]])

B = np.array([1, 2, 3])

# Add B to A
print(A+B)

# subtract B - A
print(A-B)

# Multiply A by B
print( A*B)

# Divide A by B 
print(A/B)

# Add 10 to every element of A without using another array 
print(10 + A)

# Explain why NumPy allows operations between A and B, even though their shapes are different.
# = Due to broadcasting , here B is broadcast to become the shape of (3,3) so the shape of A and B become shape allowing operations
```

#### Aggregation and Axis Problem 

```python
# Given 

arr = np.array([[10, 20, 30],
                [40, 50, 60],
                [70, 80, 90]])

# Calculate the sum of all elements
print(np.sum(arr))

# Calculate the sum of each column
print(np.sum(arr,axis=0))

# Calculate the sum of each row 
print(np.sum(arr,axis=1))

# Calculate the mean of each column 
print(np.mean(arr,axis=0))

# Calculate the maximum value of each row. 
print(np.max(arr,axis=1))

# Calculate the minimum value of each column 
print(np.min(arr,axis=0))

# Calculate the standard deviation of the entire array
print(np.std(arr))
```

#### Boolean Indexing and Filtering Problem

```python
# Given 
arr = np.array([12, 45, 67, 23, 89, 34, 90, 56, 78, 15, 100, 42])

# Extract all the elements greater than 50.
print(arr[arr>50])

# Extract all even numbers. 
print(arr[arr%2==0])

# Extract all numbers between 30 and 80 (inclusive)
print(arr[(arr>=30 )& (arr<=80)])

# Replace all elements greater than 75 with 0 
arr1=arr.copy()
arr1[arr1>75]=0
print(arr1)

# Count how manu elements are greater than 50
print(np.size(arr[arr>50]))

# Extract all elements that are either less than 20 or greater than 80.
print(arr[(arr>80) | (arr<20)])
```

#### Sorting and Searching Problem 

```python
# Given 

arr = np.array([45, 12, 89, 34, 67, 23, 90, 56, 12, 78])

# Sort the array in ascending order 
print(np.sort(arr))

# Sort the array in descending order 
print(np.sort(arr)[::-1])

# Find the indices of elements in ascending sorted order 
print(np.argsort(arr))

# Find the index of the maximum element
print(np.argsort(arr)[-1])
print(np.argmax(arr))

# Find the index of the minimum element
print(np.argsort(arr)[0])
print(np.argmin(arr))

# Find all indices where the value is 12.
print(np.where(arr==12))

# Fina the unique elements in the array 
print(np.unique(arr))
```
