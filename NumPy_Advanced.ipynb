## Broadcasting 
The term boardcasting describes how NumPy treats arrays with different shapes duing arithmetic operations.
The smaller arrays is 'broadcast' across the largest array so that they have compatile shapes. 

### Rules of broadcasting 
1. Make the two arrays have the same number of dimensions.

-> If the numbers of dimensions of the two arrays are different, add new dimensions with size 1 to the head(left) of the array with the smaller dimensions.

2. Make each dimension of the two arrays the same size
   
   -> If the size of the each dimension of the two arrays do not match , dimensions with size 1 are stretched to the size of the other array.
   
   -> If there is a dimension whose size is not 1 in either of the two arrays, it cannot be broadcasted and an error is arise.
   

```python
# Diff shapes 

a= np.arange(1,7).reshape(2,3)
b=np.arange(1,4).reshape(1,3)

print(a,b)

print(a+b)
```

# Working with mathematical formula 

```python
a=np.arange(10)
print(np.sin(a))
```

```python
# Sigmoid

def sigmoid(array):
    return 1/(1+np.exp(-(array)))

a=np.arange(1000)

print(sigmoid(a))
```

```python
# Mean square error 

actual=np.random.randint(1,50,25)
predicted=np.random.randint(1,50,25)

print(actual,predicted)
```

```python
def mse(actual,predicted):
    return np.mean((actual-predicted)**2)
print(mse(actual,predicted))
```

#### How to deal with missing Value

```python
# nan is the missing value 
a=np.array([1,2,np.nan,3,np.nan])
print(a)
```

```python
# isnan() check for missing value 

print(a[~ np.isnan(a)])
```
